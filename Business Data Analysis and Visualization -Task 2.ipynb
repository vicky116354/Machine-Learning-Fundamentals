```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Finance_Analytics_Dataset.csv")
df.head()
```

```python
#Q1
df["Category"].value_counts().plot(kind="bar")
plt.title("Transactions by Category")
plt.show()
```

```python
#Q2
sns.countplot(x="PaymentMode", data=df)
plt.title("Number of Transactions for Each Payment Mode")
plt.show()
```

```python
#Q3
df["TransactionType"].value_counts().plot(kind="bar")
plt.title("Debit vs Credit Transactions")
plt.show()
```

```python
#Q4
df.groupby("TransactionType")["Amount"].sum().plot(kind="bar")
plt.title("Debit vs Credit Amount")
plt.xlabel("Transaction Type")
plt.ylabel("Total Amount")
plt.show()
```

```python
#Q5
df.groupby("AccountType")["Amount"].sum().plot(kind="bar")
plt.title("Spending Pattern by Account Type")
plt.xlabel("Account Type")
plt.ylabel("Total Amount")
plt.show()
```

```python
#Q6
df["PaymentMode"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Payment Mode Usage Distribution")
plt.ylabel("")
plt.show()
```

```python
#Q7
sns.boxplot(x="Category", y="Amount", data=df)
plt.title("Transaction Amount by Category")
plt.xlabel("Category")
plt.ylabel("Amount")
plt.show()
```
